
6. Anda dapat menggunakan model pipeline untuk mempermudah perkejaan Anda.


```python
#membuat Pipeline (StandardScaler -> SelectKBest -> LogisticRegression)
pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('select_k', SelectKBest(score_func=f_classif, k=10)),
    ('model', LogisticRegression(random_state=42)),
])

#pelatihan model
pipeline.fit(X_train, y_train)

#evaluasi model
y_pred = pipeline.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)

print(f'Akurasi Pipeline: {accuracy * 100:.2f}%\n')
print(classification_report(y_test, y_pred))
```
