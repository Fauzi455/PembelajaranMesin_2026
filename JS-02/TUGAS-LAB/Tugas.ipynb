#Load Data

```
dpath = 'wbc.csv'
df = pd.read_csv(dpath)
df.head(10)
```

#1. Pisahkan antara variabel yang dapat digunakan dan variabel yang tidak dapat digunakan.

```
#menghapus kolom id dan unnamed:32
cols_to_drop = ['id', 'Unnamed: 32']
df = df.drop(
    columns=[col for col in cols_to_drop if col in df.columns], errors='ignore'
)
```

```
df.head(10)
```

#2. Lakukan proses encoding pada kolom "diagnosis".

```
#menampilkan nilai unik pada kolom diagnosis
print("Nilai unik:", df['diagnosis'].unique())
```

```
#mengubah nilai kategorial 'M' dan 'B' menjadi 1 dan 0 pada diagnosis
le = LabelEncoder()
df['diagnosis'] = le.fit_transform(df['diagnosis'])
```

```
df.head()
```

#3. Lakukan proses standardisasi pada semua kolom yang memiliki nilai numerik.

```
std = StandardScaler()

# Daftar seluruh 30 kolom fitur numerik
features = [
    'radius_mean',
    'texture_mean',
    'perimeter_mean',
    'area_mean',
    'smoothness_mean',
    'compactness_mean',
    'concavity_mean',
    'concave points_mean',
    'symmetry_mean',
    'fractal_dimension_mean',
    'radius_se',
    'texture_se',
    'perimeter_se',
    'area_se',
    'smoothness_se',
    'compactness_se',
    'concavity_se',
    'concave points_se',
    'symmetry_se',
    'fractal_dimension_se',
    'radius_worst',
    'texture_worst',
    'perimeter_worst',
    'area_worst',
    'smoothness_worst',
    'compactness_worst',
    'concavity_worst',
    'concave points_worst',
    'symmetry_worst',
    'fractal_dimension_worst',
]
df[features] = std.fit_transform(df[features])
```
